<font size = "3">

- Suppose we want to see how many experiments it takes to get a **four successes** (in total, not consecutively). Complete the code below so that it simulates this sequence of experiments. After your code completes, the variable `num_trials` should be equal to the number of experiments it took to achieve the fourth "Success". 

- **Hint:** We've added some code that helps prevents an infinite loop. Remove it at your own peril...

```python
### SOLUTION
import numpy as np

num_success = 0
num_trials = 0
this_prevents_infinite_loop = 0

while num_success < 4:
    outcome = np.random.choice(["Success", "Failure"], p = [0.45, 0.55])

    num_trials += 1

    if outcome == "Success":
        num_success += 1

    this_prevents_infinite_loop = this_prevents_infinite_loop + 1
    if this_prevents_infinite_loop == 100000:
        print("The code shouldn't be taking this long. Try again...")
        break



print("Number of trials:", num_trials)
```

**Question (F)** Conditionally removing elements from a set

<font size = "3">

The code cell below does the following:

- Defines a tuple `course_numbers` containing the course numbers of some courses offered at Emory.

- Defines a tuple `course_name` containing the names of the corresponding courses, based on index. (Since "HIST 267" is stored in course_numbers[2], the name of this course is "The Civil Rights Movement" because that is what is stored in course_name[2])

- Randomly chooses one of the course names, saving it to the variable `name_to_remove`.

- Randomly chooses one of six course schedules, saving it to the variable `course_schedule` (which is a **set**)

Complete the code so that it:

1. Finds the index of the course name corresponding to "name_to_remove"

2. Uses the index to find the course number corresponding to this course name.

3. Checks if this course number belongs to the set "course_schedule", and if so, removes it from the list.


```python
###### Set-up code
import random

# Define course numbers
course_numbers = ("DATASCI 185", "CL 102", "HIST 267", "CHEM 333", "DATASCI 325", "HIST 247", "HIST 360", "ENG 205")

# Define corresponding course names
course_name = ("Computational Methods", "Classical Mythology", "The Civil Rights Movement", 
               "Biophysical Chemistry", "Evolutionary Game Theory", "Napoleon's Europe", "History of Mexico", "Poetry")


# Randomly choose name of course
name_to_remove = random.choice(course_name)

# Randomly choose one of these 6 course schedules
course_schedule_a = {"DATASCI 185", "CL 102", "HIST 267", "CHEM 333"}
course_schedule_b = {"DATASCI 185", "DATASCI 325", "HIST 247", "CHEM 333"} 
course_schedule_c = {"DATASCI 185", "CL 102", "HIST 360", "CHEM 333"} 
course_schedule_c = {"DATASCI 325", "CL 102", "HIST 360", "CHEM 333"} 
course_schedule_d = {"DATASCI 185", "CL 102", "ENG 205", "CHEM 333"} 
course_schedule_e = {"DATASCI 325", "HIST 360", "HIST 247", "HIST 267"}
schedule_list = [course_schedule_a, course_schedule_b, course_schedule_c, course_schedule_d, course_schedule_e]

course_schedule = random.choice(schedule_list)

print(course_schedule)
print(name_to_remove)
###### (end of set-up code)


#### SOLUTION
idx = course_name.index(name_to_remove)
course_no = course_numbers[idx]

if course_no in course_schedule:
    course_schedule.remove(course_no)

print(course_schedule)

```
